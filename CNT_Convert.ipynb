```python
import os
import numpy as np
import mne
```

```python
def convert_trec_cnt_to_set(study_num, subj_num, session,
                             base_dir='/Users/elizabethkaplan/Desktop/Greene/subject_folders/'):
    """
    Example:
        convert_trec_cnt_to_set(353, 12415, 'pre')
        convert_trec_cnt_to_set('353', '12415', 'post')
    """
    study_num = str(study_num)
    subj_num = str(subj_num)
    session = str(session).lower()

    if session not in ('pre', 'post'):
        raise ValueError('Session must be "pre" or "post".')

    subj_id = f'TReC_{subj_num}'
    subj_dir = os.path.join(base_dir, subj_id)

    # Filename structure
    base_filename = f'study-{study_num}_sub-{subj_num}_task-eeg_ses-{session}'

    # Capitalizes the session; pre -> Pre
    session_cap = session[0].upper() + session[1:]
    base_filename2 = f'TReC_{subj_num}_{session_cap}'

    input_file = os.path.join(subj_dir, base_filename + '.cnt')
    output_file = base_filename2 + '.set'

    if not os.path.isfile(input_file):
        raise FileNotFoundError(f'Input file not found:\n{input_file}')

    raw = mne.io.read_raw_ant(input_file, preload=True)  # replaces pop_loadeep_v4

    if np.isnan(raw.get_data()).any():
        raise ValueError('NaNs after ANT import')

    raw.export(os.path.join(subj_dir, output_file), fmt='eeglab', overwrite=True)

    print(f'Saved:\n{os.path.join(subj_dir, output_file)}')

    return raw
```
